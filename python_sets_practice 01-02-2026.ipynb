### Q8
Count unique elements in:
```python
nums = [1, 2, 2, 3, 4, 4, 5]
```

````python
nums = [1, 2, 2, 3, 4, 4, 5]
unique = []
for num in nums:
    if num not in unique:
        unique.append(num)

print("Count of unique elements:", len(unique))
````
